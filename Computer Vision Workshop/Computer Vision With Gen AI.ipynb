### Whenever a new image is created in another window, user has to close it manuaally. 

### I do not want to do that. 

```python
import cv2

my_img = cv2.imread(r"C:\Users\artem\OneDrive - Indian Oil Corporation Limited\Documents\Gen AI and Agenetic AI\2. Computer Vision with Gen AI\rabbit.jpg")

while True:
    cv2.imshow('my image', my_img)
    
    if cv2.waitKey(1) & 0xFF == 27:
        break
        
cv2.destroyAllWindows()

# Here popup will close if we press the escape key
```
