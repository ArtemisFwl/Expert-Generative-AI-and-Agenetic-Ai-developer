## Spam Ham Classification using NLTK - TFIDF

```python
import pandas as pd
```

```python
df=pd.read_csv('spam.csv', encoding="latin1" )
```

```python
df=df.iloc[:, :2]
df.columns=['labels', 'message']
```

```python
df.head()
```

```python
import re
import nltk 
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
corpus=[]

for i in range(0, len(df)):
    review=re.sub('[^a-zA-z]', ' ', df['message'][i])
    review=review.lower()
    review=review.strip()
    review=review.split()

    review= [ps.stem(words) for words in review if not words in stopwords.words('english')]
    review= ' '.join(review)
    corpus.append(review)
```

### Creating tf-idf model

```python
from sklearn.feature_extraction.text import TfidfVectorizer 
tv= TfidfVectorizer(max_features=2500, ngram_range=(1,2))
X=tv.fit_transform(corpus).toarray()
```

```python
X
```

```python
tv.vocabulary_
```

### Train Test SPlit 

```python
y=pd.get_dummies(df['labels'])
y = y.astype(int)
y=y.iloc[:, 0].values
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2)
```

```python
from sklearn.naive_bayes import MultinomialNB
spam_tfidf_model=MultinomialNB()
spam_tfidf_model.fit(X_train, y_train)
```

```python
y_pred=spam_tfidf_model.predict(X_test)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```
