```python
paragraph='''
My dear young friends,
I am delighted to be with you today. When I see you, I see the future of our nation. Each one of you has a dream. A dream is not what you see in sleep, but a dream is something that does not let you sleep.

You have to dream before your dreams can come true. If you want to shine like the sun, first burn like the sun. Great dreams of great dreamers are always transcended.

We should not give up and we should not allow the problem to defeat us. Difficulties in life are necessary because they give us the opportunity to learn and to grow.

Excellence happens not by accident. It is a process. It requires hard work, perseverance, learning, studying, sacrifice and most of all, love of what you are doing.

Let us sacrifice our today so that our children can have a better tomorrow.

My message, especially to young people, is to have courage to think differently, to invent, to travel the unexplored path, to discover the impossible and to conquer the problems.

This is your mission. This is your responsibility. This is your life.'''
```

```python
from nltk.corpus import stopwords
```

```python
# we have to apply stemming and remove storwords from the paragraph
```

```python
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()
```

```python
from nltk.tokenize import sent_tokenize
```

```python
sentences=sent_tokenize(paragraph)
```

```python
type(sentences)
```

```python
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()
```

```python
import nltk
```

```python
# Apply stopwords and filter and apply stemming
for i in range(len(sentences)):
    words= nltk.word_tokenize(sentences[i])
    words=[stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i]= " ".join(words)
```

```python
from nltk.stem import SnowballStemmer
sb_stemmer = SnowballStemmer("english")
```

```python
sentences=sent_tokenize(paragraph)
```

```python
# Apply stopwords and filter and apply stemming
for i in range(len(sentences)):
    words= nltk.word_tokenize(sentences[i])
    words=[sb_stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i]= " ".join(words)
```

```python
sentences
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
```

```python
# Apply stopwords and filter and apply stemming
for i in range(len(sentences)):
    words= nltk.word_tokenize(sentences[i])
    words=[lemmatizer.lemmatize(word, pos='v') for word in words if word not in set(stopwords.words('english'))]
    sentences[i]= " ".join(words)
```
